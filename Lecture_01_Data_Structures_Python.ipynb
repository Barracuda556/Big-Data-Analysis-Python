## 1. Векторы

В R вектор — основная одномерная структура данных.

### Эквиваленты в Python:
| R | Python |
|---|--------|
| `c()`, `seq()` | `list`, `range()`, `np.arange()`, `np.linspace()`, `pd.Series` |
| Индексация с 1 | Индексация с **0** |
| `V[5:9]` | `V[4:9]` (срез) или `V[[4,5,6,7,8]]` |

### Основные способы создания «векторов»:


```python
# 1.1 Создание вектора S1 со значениями от 1 до 10

# Способ 1: обычный список Python
S1_list = list(range(1, 11))          # range(1, 11) → 1..10
print("Список:", S1_list)

# Способ 2: NumPy-массив (предпочтительно для вычислений)
S1 = np.arange(1, 11)                 # аналог seq(1, 10)
print("NumPy array:", S1)
print("Тип:", type(S1))

# Способ 3: Pandas Series (удобно, если нужны метки/имена)
S1_series = pd.Series(range(1, 11))
print("Pandas Series:\n", S1_series)
```

```python
# 1.2 Вектор S2 от 1 до 10 с шагом 0.5
# В R: seq(from=1, to=10, by=0.5)

S2 = np.arange(1, 10.5, 0.5)          # stop не включается, поэтому 10.5
# или
S2_alt = np.linspace(1, 10, num=19)   # 19 точек от 1 до 10 включительно

print("S2 (arange):", S2)
print("Длина S2:", len(S2))
```

```python
# 1.3 Объединение векторов S1 и S2 в вектор S
# В R: c(S1, S2)

S = np.concatenate([S1, S2])
# альтернативы:
# S = np.hstack([S1, S2])
# S = np.r_[S1, S2]

print("Объединённый вектор S:")
print(S)
print("Длина S:", len(S))
```

```python
# 1.4 Вывести 3-е, 4-е и 5-е значения вектора S
# В R индексация с 1 → S[3:5]
# В Python индексация с 0 → S[2:5]

print("Элементы 3, 4, 5:", S[2:5])

# или явно:
print("Через fancy indexing:", S[[2, 3, 4]])
```

```python
# 1.5 Выбрать только первое и четвёртое значения вектора S
# В R: S[c(1,4)]
# В Python: S[[0, 3]]

print("1-й и 4-й элементы:", S[[0, 3]])
```

### Полезные методы и атрибуты NumPy-массивов / списков
- `len(arr)` / `arr.shape` — длина / размерность
- `arr.dtype` — тип данных
- `arr[start:stop:step]` — срезы
- `np.append()`, `np.concatenate()`, `np.hstack()`, `np.vstack()`
- `arr.min()`, `arr.max()`, `arr.mean()`, `arr.sum()` и т.д.

## 2. Таблицы данных (DataFrame)

В R основная табличная структура — `data.frame`.

В Python — **`pandas.DataFrame`**.

Это самый важный объект для анализа данных.


```python
# 2.1 Создание векторов (столбцов)
city = ["City1", "City1", "City2", "City2", "City3", "City3"]
sex = ["Male", "Female", "Male", "Female", "Male", "Female"]
number = [12450, 10345, 5670, 5800, 25129, 26000]

print(city)
print(sex)
print(number)
```

```python
# 2.3 Создание таблицы CITY с помощью pd.DataFrame
# В R: data.frame(City=city, Sex=sex, Number=number)

CITY = pd.DataFrame({
    "City": city,
    "Sex": sex,
    "Number": number
})

CITY
```

```python
# 2.4 Просмотр структуры таблицы
# В R: str(CITY)
# В Python: .info() или .dtypes + .shape

print("Информация о таблице:")
CITY.info()

print("\nТипы столбцов:")
print(CITY.dtypes)

print("\nРазмерность (строки, столбцы):", CITY.shape)
```

```python
# 2.5 Имена переменных (столбцов)
# В R: names(CITY)
# В Python: CITY.columns

print(CITY.columns.tolist())
# или просто
CITY.columns
```

```python
# 2.6 Извлечь элементы 1–3 из столбца Number
# В R: CITY$Number[1:3]
# В Python (индексация с 0): CITY["Number"].iloc[0:3]

print(CITY["Number"].iloc[0:3])

# альтернативы:
# CITY.Number.iloc[0:3]
# CITY.loc[0:2, "Number"]   # .loc использует метки (здесь совпадают с индексами)
```

```python
# 2.7 Все значения численности, превышающие 10000
# В R: CITY$Number[CITY$Number > 10000]

print(CITY[CITY["Number"] > 10000])

# только столбец Number:
print("\nТолько Number > 10000:")
print(CITY.loc[CITY["Number"] > 10000, "Number"])
```

```python
# 2.8 Все значения численности мужского населения
# В R: CITY$Number[CITY$Sex == "Male"]

print(CITY[CITY["Sex"] == "Male"])

# или только числа:
print("\nЧисленность мужчин:")
print(CITY.loc[CITY["Sex"] == "Male", "Number"])
```

```python
# 2.9 Первые и последние строки
# В R: head(CITY, n=3), tail(CITY, n=2)

print("Первые 3 строки:")
display(CITY.head(3))

print("\nПоследние 2 строки:")
display(CITY.tail(2))
```

### Ключевые методы и атрибуты DataFrame
| Метод / атрибут | Описание |
|-----------------|----------|
| `.head(n)`, `.tail(n)` | первые / последние n строк |
| `.info()` | структура, типы, пропуски |
| `.describe()` | описательная статистика |
| `.shape`, `.columns`, `.index` | размер, имена столбцов, индексы |
| `.loc[]`, `.iloc[]` | доступ по меткам / по позициям |
| `df[условие]` | фильтрация строк |
| `df["col"]` или `df.col` | выбор столбца (Series) |
| `pd.concat()`, `pd.merge()` | объединение таблиц |

## 3. Матрицы

В R матрица — двумерный массив чисел.

В Python основной инструмент — **`numpy.ndarray`** с `ndim == 2`.


```python
# 3.1 Создание матрицы размерности 2×8
# В R: m1 <- seq(1, 16); dim(m1) <- c(2, 8)

m1 = np.arange(1, 17).reshape(2, 8)   # 16 элементов → 2 строки × 8 столбцов
print("Матрица 2×8:")
print(m1)
print("Форма:", m1.shape)
```

```python
# 3.2 Создание матрицы 4×6 с помощью np.array / np.matrix / reshape
# В R: matrix(seq(1, 24), nrow=4, ncol=6)

m2 = np.arange(1, 25).reshape(4, 6)
print("Матрица 4×6:")
print(m2)

# Можно указать порядок заполнения (по умолчанию по строкам, как в C):
# m2 = np.arange(1, 25).reshape(4, 6, order='F')  # по столбцам (как в Fortran/R)
```

```python
# 3.3 Транспонирование матрицы
# В R: t(m2)

m3 = m2.T          # или np.transpose(m2)
print("Транспонированная матрица (6×4):")
print(m3)
print("Форма:", m3.shape)
```

```python
# 3.4 Квадратная матрица и определитель
# В R: det(matrix(...))

square = np.array([[1, 2],
                   [3, 4]])
print("Квадратная матрица:")
print(square)

det = np.linalg.det(square)
print("Определитель:", det)
```

### Полезные операции с матрицами NumPy
- `A @ B` или `np.dot(A, B)` — матричное умножение
- `A * B` — поэлементное умножение
- `np.linalg.inv(A)` — обратная матрица
- `np.linalg.eig(A)` — собственные значения и векторы
- `A.reshape(...)`, `A.ravel()`, `A.flatten()`

## 4. Многомерные массивы

В R `array()` позволяет создавать массивы любой размерности.

В Python — тот же `numpy.ndarray`.


```python
# 4.1 Массив из 24 элементов
# В R: array(1:24, dim=c(3, 2, 4)) или подобное

# Создадим массив размерности (3, 2, 4) — 3×2 матрицы, 4 «слоя»
M = np.arange(1, 25).reshape(3, 2, 4)
print("Массив M формы", M.shape)
print(M)

# Альтернатива: явно через np.array и reshape
```

## 5. Списки

В R список (`list`) может содержать элементы разных типов и длин.

В Python:
- обычный `list` — упорядоченная коллекция
- `dict` — именованная коллекция (часто ближе к R-списку с именами)


```python
# 5.1 Создание списка L1: числа от 1 до 6 + TRUE/FALSE три раза
# В R: list(1:6, rep(c(TRUE, FALSE), 3))

numbers = list(range(1, 7))
booleans = [True, False] * 3

L1 = [numbers, booleans]          # обычный список списков
# или
L1_dict = {
    "integers": numbers,
    "booleans": booleans
}

print("L1 (list):")
print(L1)

print("\nL1_dict:")
print(L1_dict)
```

```python
# 5.2 Присвоение имён элементам списка
# В R: names(L1) <- c("integers", "booleans")

# Для словаря имена уже есть.
# Для обычного списка можно использовать словарь или namedtuple,
# но чаще всего в анализе данных используют dict или pd.Series.

L1_named = {
    "integers": list(range(1, 7)),
    "booleans": [True, False, True, False, True, False]
}

print(L1_named)
print("\nКлючи:", list(L1_named.keys()))
print("Значения integers:", L1_named["integers"])
```

## 6. Факторы (категориальные переменные)

В R `factor` — специальный тип для категориальных данных с фиксированными уровнями (levels).

В Python / Pandas — **`pandas.Categorical`** или столбец с `dtype="category"`.


```python
# 6.1 Вектор из 6 объектов трёх классов: Yes, No, Perhaps → фактор

answers = ["Yes", "No", "Perhaps", "Yes", "No", "Perhaps"]

# Создание Categorical
f1 = pd.Categorical(answers, categories=["Yes", "No", "Perhaps"])
print(f1)
print("Уровни (categories):", f1.categories.tolist())
print("Коды:", f1.codes)
```

```python
# Использование в DataFrame
df_cat = pd.DataFrame({"answer": answers})
df_cat["answer"] = df_cat["answer"].astype("category")
print(df_cat.dtypes)
print(df_cat["answer"])
```

### Зачем нужны категории?
- Экономия памяти
- Возможность задать порядок (ordered categorical)
- Корректная работа многих статистических и ML-методов

## 7. Фрейм данных (ещё раз про DataFrame)

Пункт 7 задания — создание data.frame из разнотипных векторов, включая factor.


```python
# 7.1 Создание фрейма F1

b = ["a", "b", "c", "a"]
d = [(i % 2 == 0) for i in range(1, 5)]          # True/False
e = pd.Categorical(["soft", "hard", "soft", "medium"],
                   categories=["soft", "medium", "hard"])

F1 = pd.DataFrame({
    "b": b,
    "d": d,
    "e": e
})

print(F1)
print("\nТипы столбцов:")
print(F1.dtypes)
```

## Краткая шпаргалка: R → Python

| Задача в R | Python (NumPy / Pandas) |
|------------|--------------------------|
| `c(1,2,3)` | `[1,2,3]` или `np.array([1,2,3])` |
| `seq(1,10, by=0.5)` | `np.arange(1, 10.5, 0.5)` |
| `data.frame(...)` | `pd.DataFrame({...})` |
| `matrix(..., nrow, ncol)` | `np.arange(...).reshape(nrow, ncol)` |
| `array(..., dim=...)` | `np.arange(...).reshape(...)` |
| `list(...)` | `list(...)` или `dict(...)` |
| `factor(...)` | `pd.Categorical(...)` / `.astype("category")` |
| `str(df)` | `df.info()` |
| `names(df)` | `df.columns` |
| `df$col` | `df["col"]` или `df.col` |
| `df[df$col > 10]` | `df[df["col"] > 10]` |
| `head(df, 3)` | `df.head(3)` |
| `t(m)` | `m.T` |
| `det(m)` | `np.linalg.det(m)` |
| Индексация с 1 | Индексация с **0** |
